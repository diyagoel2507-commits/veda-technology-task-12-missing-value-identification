```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
missing_values = missing_values[missing_values > 0]

print("Columns with missing values:")
print(missing_values)
```

```python
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_count,
    "Missing Percentage": missing_percentage
})

missing_summary = missing_summary[missing_summary["Missing Values"] > 0]

missing_summary
```

```python
import matplotlib.pyplot as plt

missing_summary["Missing Values"].plot(
    kind="bar",
    figsize=(8, 5),
    title="Missing Values by Column"
)

plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Short Findings

- The dataset contains missing values in the Age, Cabin, and Embarked columns.
- The Age column has 177 missing values.
- The Cabin column has 687 missing values, which is the highest among all columns.
- The Embarked column has only 2 missing values.
- Cabin contains a very large proportion of missing data and should not be blindly deleted without understanding its impact.
- Age missing values can be handled using an appropriate imputation method such as median imputation.
- Embarked has very few missing values and can be handled after checking the most suitable value.
- No rows were deleted because missing values should be treated based on the purpose and context of the analysis.

## Interview Questions

### 1. What is a missing value?

A missing value is a data field for which no valid value has been recorded.

### 2. Why can blindly deleting rows be risky?

Blindly deleting rows can reduce the dataset size and may remove useful information. It can also introduce bias into the analysis.
